```python
import matplotlib.pyplot as plt

# The experimental data
sample_sizes = ['1k', '10k', '50k']
fid_scores = [73.0, 32.0, 29.0]

# Setup the canvas
plt.figure(figsize=(8, 5))

# Plot the line chart with markers
plt.plot(sample_sizes, fid_scores, marker='o', markersize=8, linestyle='-', color='#1f77b4', linewidth=2.5)

# Add the exact FID values as text labels above the markers
for i, score in enumerate(fid_scores):
    plt.text(i, score + 2.5, str(score), ha='center', fontweight='bold', fontsize=11)

# Academic formatting
plt.title('Impact of Evaluation Sample Volume on FID Score', fontsize=14, pad=15)
plt.xlabel('Number of Generated Samples', fontsize=12)
plt.ylabel('Fréchet Inception Distance (FID)', fontsize=12)

# Add a subtle grid for readability
plt.grid(True, linestyle='--', alpha=0.6)
plt.ylim(20, 85) 

# Save to disk with high resolution for Typst
plt.savefig('6.2.4_fid.png', dpi=300, bbox_inches='tight')
print("Successfully saved 6.2.4_fid.png")
```
